```python
import numpy as np

temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temparatures:")
print(temperatures_w1)
```

```python
# Shape,Size,Datatype

print("Shape of the array:",temperatures_w1.shape)
print("Data type of the array:",temperatures_w1.dtype)
print("Size  of the array:",temperatures_w1.size)
```

```python
# Convert Celsius to Fahrenheit
temperatures_f=(temperatures_w1 * 9/5) + 32
print("Temperature in Fahrenheit:")
print(temperatures_f)
```

```python
# first 3 days,last 2 days,middle 3

first_three=temperatures_w1[0:3]
weekend = temperatures_w1[-2:]
middle_three=temperatures_w1[2:5]

print("First three days:", first_three)
print("Weekend temperatures(Last two days):", weekend)
print("Middle three days:", middle_three)
```

```python
# 2D creation

temperatures=np.array([
        [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
        [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperature Data:")
print(temperatures)
print("-------------------------------------------------")

print("Shape of the array:",temperatures.shape)
print("Datatype of the array:",temperatures.dtype)
print("Total number of elements:",temperatures.size)

# week 1 and 2 Extract
print("-------------------------------------------------")
week1 = temperatures[0,:]
week2=temperatures[1,:]

print("Week 1 Temperatures:",week1)
print("Week 2 Temperatures:",week2)

weekend=temperatures[:,-2:]
print("Weekend Temperatures:")
print(weekend)
```

# Pandas

```python
import pandas as pd

marks=pd.Series(
    [95,92,89,85,80],
    index=['Rank1','Rank2','Rank3','Rank4','Rank5']
)

print(marks)
```

```python
first_mark=marks.iloc[0]
print("First Rank Mark:",first_mark)

top_three=marks.loc[['Rank1','Rank2','Rank3']]
print("\nTop 3 Ranks:")
print(top_three)

third_rank=marks.iloc[2]
print("\nThird Rank Mark:",third_rank)

high_marks=marks[marks>90]
print("\nRanks with marks greater than 90:")
print(high_marks)
```

```python
# Modify Rank1 to 100

marks.loc['Rank1']=100
print("Updated Marks:")
print(marks)
```

```python
# Remove the last mark

marks=marks.drop('Rank5')

print("Marks after removing last rank:")
print(marks)
```

```python
# cgpa

cgpa=marks/10

print("CGPA:")
print(cgpa)
```

# DataFrame

```python
transactions=pd.DataFrame({
    'TransactionID':[101,102,103,104,105,106,107,108,109,110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'],

    'Region':[
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'],
    
    'Amount':[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

```python

# Display the complete DataFrame
print("Complete DataFrame:")
print(transactions)
print("__________________________________________________________________________")


# Display first 5 rows
print("\n First 5 Rows:")
print(transactions.head())
print("__________________________________________________________________________")

# Display last 5 rows
print("\n Last 5 Rows:")
print(transactions.tail())
print("__________________________________________________________________________")

# Display shape
print("\n Shape:")
print(transactions.shape)
print("__________________________________________________________________________")

# Display column Name
print("\n Column Name:")
print(transactions.columns)
print("__________________________________________________________________________")

# Display data types
print("\n Data Types:")
print(transactions.dtypes)
print("__________________________________________________________________________")


# Selected Columns
selected_columns=transactions[['ProductCategory','Amount']]
print(selected_columns)
print("__________________________________________________________________________")


#last 3 colomns
last_three=transactions.iloc[:,-3:]
print("Last 3 columns:")
print(last_three)
print("__________________________________________________________________________")


# Filtered data
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]
print("Filtered Transactions:")
print(filtered_data)
print("__________________________________________________________________________")


# Unique
unique_regions = transactions['Region'].unique()

print("Unique Regions:")
print(unique_regions)
print("__________________________________________________________________________")



# Mean
region_mean = transactions.groupby('Region')['Amount'].mean()

print("Mean Amount for Each Region:")
print(region_mean)

```
